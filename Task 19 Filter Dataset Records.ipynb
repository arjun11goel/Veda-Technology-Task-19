# Level 1 • Day 19: Filter Dataset Records
**Objective:** Learn how to retrieve relevant records from a dataset using one or more conditions with Pandas.

In this notebook, we will load the Titanic dataset and explore various ways to filter rows based on single and multiple conditions using Boolean indexing, logical operators, and Pandas methods.

```python
# Import necessary libraries
import pandas as pd
import seaborn as sns

# Load the Titanic dataset directly from the online source or via seaborn
# You can also use pd.read_csv('titanic.csv') if you downloaded it locally.
df = sns.load_dataset('titanic')

# Display the first 5 rows of the dataset to understand its structure
print("Dataset Shape:", df.shape)
df.head()
```

### Example 1: Filtering with a Single Condition
**Explanation:** We filter the DataFrame to find all passengers who are in First Class (`pclass == 1`). We use a simple Boolean condition `df['pclass'] == 1` inside the bracket notation `df[...]` to extract only the matching rows.

```python
# Filter passengers in First Class (pclass == 1)
first_class_passengers = df[df['pclass'] == 1]

print(f"Total First Class passengers: {len(first_class_passengers)}")
first_class_passengers.head()
```

### Example 2: Filtering with Multiple Conditions (`AND` & `OR`)
**Explanation:** When combining multiple conditions in Pandas:
* The bitwise AND operator (`&`) is used to satisfy **all** conditions.
* The bitwise OR operator (`|`) is used to satisfy **at least one** condition.
* **Important:** Parentheses `()` are mandatory around each individual condition because Python evaluates bitwise operators (`&`, `|`) with higher precedence than comparison operators (`==`, `>`, `<`).

Here, we filter for female passengers (`sex == 'female'`) who were in First Class (`pclass == 1`) and survived (`survived == 1`).

```python
# Filter: Female passengers in First Class who survived
# Parentheses are required around each condition!
survived_first_class_females = df[(df['sex'] == 'female') & (df['pclass'] == 1) & (df['survived'] == 1)]

print(f"Total matching records: {len(survived_first_class_females)}")
survived_first_class_females.head()
```

### Example 3: Filtering Using `.isin()` Method
**Explanation:** The `.isin()` method is a cleaner, more readable alternative when checking if column values match any value in a list (acting as a clean alternative to multiple `|` OR conditions). Here, we filter passengers who embarked from either port 'C' (Cherbourg) or port 'Q' (Queenstown).

```python
# Filter passengers who embarked from Cherbourg ('C') or Queenstown ('Q')
embarked_c_or_q = df[df['embarked'].isin(['C', 'Q'])]

print(f"Total matching records: {len(embarked_c_or_q)}")
embarked_c_or_q[['embarked', 'class', 'fare']].head()
```

## Interview Questions & Answers

### 1. How do you filter rows in Pandas?
* **Answer:** You can filter rows in Pandas using **Boolean Indexing** by passing a conditional expression inside square brackets (e.g., `df[df['age'] > 30]`). Alternatively, you can use the `.loc[]` indexer (e.g., `df.loc[df['age'] > 30]`), which is especially useful when filtering both rows and specific columns simultaneously.

### 2. How do you combine multiple conditions?
* **Answer:** You combine multiple conditions using bitwise operators:
  * **AND (`&`)**: All conditions must be true.
  * **OR (`|`)**: At least one condition must be true.
  * **NOT (`~`)**: Negates the condition.
  * *Note:* Each individual condition must be wrapped in parentheses `()`. You can also use the `.isin()` method for checking multiple values against a list or the `.query()` method for writing SQL-like query strings.

### 3. Why are parentheses important in Pandas conditions?
* **Answer:** Parentheses are mandatory because Python's operator precedence rules give bitwise operators (`&`, `|`) higher priority than comparison operators (`==`, `>`, `<`, etc.). Without parentheses, Python would attempt to perform bitwise operations on the raw column expressions before evaluating the comparisons, resulting in a `ValueError` or incorrect logical evaluation.
